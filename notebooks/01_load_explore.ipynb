## 📓 Notebook 01: Load & Explore

**Skills practiced:** `.info()`, identifying the target column, a plain `train_test_split`, a first baseline model with RMSE.

**Goal:** Load the bike-share data and record the Stage 1 baseline RMSE (raw numeric features only) that every later stage must beat.

```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error

RANDOM_STATE=42
```

```python
df=pd.read_csv('../data/raw/store_sharing.csv')
df.info()
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['hour'] = df['timestamp'].dt.hour
x = df.drop(columns=['timestamp', 'cnt'])
y=df['cnt']
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=RANDOM_STATE)
print(X_train.shape, X_test.shape, y_train.mean().round(3), y_test.mean().round(3))
```

```python
m=RandomForestRegressor(n_estimators=250, random_state=RANDOM_STATE)
m.fit(X_train,y_train)
p=m.predict(X_test)
```

```python
RMSE=root_mean_squared_error(y_test, p)
print(round(RMSE,2))
```

```python
X_train.to_csv('../data/processed/stage_1/x_train.csv', index=False)
X_test.to_csv('../data/processed/stage_1/x_test.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)
joblib.dump(m, '../models/stage1_model.joblib')
```
