## 📓 Notebook 04: Feature Scaling

**Skills practiced:** `StandardScaler`, `MinMaxScaler`, `.describe()` before and after, fitting scalers on train only.

**Goal:** Scale the weather columns both ways and compare the resulting statistics side by side.

```python
import pandas as pd
import joblib

from sklearn.preprocessing import StandardScaler,MinMaxScaler

from sklearn.metrics import root_mean_squared_error
from sklearn.ensemble import RandomForestRegressor
```

```python
x_train=pd.read_csv('../data/processed/stage_2/x_train_enc.csv')
x_test=pd.read_csv('../data/processed/stage_2/x_test_enc.csv')
y_train=pd.read_csv('../data/processed/y_train.csv')
y_test=pd.read_csv('../data/processed/y_test.csv')
```

```python
x_train[['t1','hum','wind_speed']].describe()
```

```python
sc=StandardScaler()
sc.fit(x_train)

X_train_scaled = sc.transform(x_train)
X_test_scaled = sc.transform(x_test)

X_train_scaled = pd.DataFrame(X_train_scaled, columns=x_train.columns, index=x_train.index)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=x_test.columns, index=x_test.index)
```

```python
X_train_scaled[['t1','hum','wind_speed']].describe()
```

```python
sc2=MinMaxScaler()
sc2.fit(x_train)

X_train_scaled2 = sc2.transform(x_train)
X_test_scaled2 = sc2.transform(x_test)


X_train_scaled2 = pd.DataFrame(X_train_scaled2, columns=x_train.columns, index=x_train.index)
```

```python
X_train_scaled2[['t1','hum','wind_speed']].describe()
```

```python
for name, (tr, te) in {
    'raw': (x_train, x_test),
    'standard': (X_train_scaled, X_test_scaled),
    'minmax': (X_train_scaled2, X_test_scaled2),
}.items():
    rf = RandomForestRegressor(n_estimators=250, random_state=42, n_jobs=-1)
    rf.fit(tr, y_train['cnt'])
    print(name, round(root_mean_squared_error(y_test['cnt'], rf.predict(te)), 2))
```

```python
X_train_scaled.to_csv('../data/processed/stage_2/x_train_sc.csv')
X_test_scaled.to_csv('../data/processed/stage_2/x_test_sc.csv')
x_train.to_csv('../data/processed/stage_2/x_train_unsc.csv')
x_test.to_csv('../data/processed/stage_2/x_test_unsc.csv')
joblib.dump(rf, '../models/stage2_model.joblib')
```
